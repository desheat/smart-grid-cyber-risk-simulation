```python
import pandas as pd
```

```python
data = {
    "Threat": [
        "Smart Meter Spoofing",
        "Smart Meter Tampering",
        "Network DoS",
        "Data Interception",
        "Control System Tampering",
        "Privilege Escalation",
        "Dashboard Spoofing",
        "Dashboard Data Leak",
        "Renewable Tampering",
        "Renewable Disruption"
    ],
    "Component": [
        "Smart Meters",
        "Smart Meters",
        "Communication Network",
        "Communication Network",
        "Grid Control System",
        "Grid Control System",
        "Monitoring Dashboard",
        "Monitoring Dashboard",
        "Renewable Energy Source",
        "Renewable Energy Source"
    ],
    "Likelihood": [4,3,5,3,4,3,3,2,3,2],
    "Operational Impact": [3,4,5,3,5,5,3,2,4,4],
    "ESG Impact": [2,2,3,2,4,4,2,3,4,4]
}

df = pd.DataFrame(data)

df
```

```python
df["Risk Score"] = df["Likelihood"] * df["Operational Impact"] * df["ESG Impact"]

df
```

```python
df = df.sort_values(by="Risk Score", ascending=False)

df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df["Threat"], df["Risk Score"])
plt.xticks(rotation=45)
plt.title("Smart Grid Cyber Risk Ranking")
plt.xlabel("Threat")
plt.ylabel("Risk Score")
plt.show()
```

Cyber Risk Analysis

The results of the risk scoring model indicate that the highest-risk threats are Control System Tampering, Network Denial of Service (DoS), and Privilege Escalation. Control System Tampering ranked as the highest risk due to its significant impact on system operations, as the control system is responsible for managing and coordinating the entire smart grid. Any unauthorized modification to this system could disrupt grid stability and lead to widespread operational issues.
Network Denial of Service also ranked highly because the communication network is essential for transmitting data between components. If the network becomes unavailable, the system cannot function effectively. Additionally, Privilege Escalation presents a critical risk because it allows unauthorized users to gain elevated access to system controls, increasing the likelihood of further system compromise.
These findings highlight that the most critical vulnerabilities exist within components responsible for system control and communication, which are essential to maintaining reliable and efficient smart grid operations.
